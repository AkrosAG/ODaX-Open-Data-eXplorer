```python
import pandas as pd
import os
```

```python
########### FIRST AND LAST DAY OF ANALYSIS AND NUMBER OF MEASUREMENTS TAKEN as well as number of nans ###########
BASE = os.path.dirname(os.path.abspath("analysis_pollutants.ipynb"))
pollutants = [f for f in os.listdir(BASE) if f.lower().endswith(".csv") and os.path.isfile(os.path.join(BASE, f))]

pollutant_unit = []
pollutant_kind = []
df_results = pd.DataFrame()
i = 0
for poll in pollutants:
    PATH_STATIONS = os.path.join(BASE,poll)
    df = pd.read_csv(PATH_STATIONS, encoding='latin-1', sep = ';', skiprows=5)
    first_day = df['Datum/Zeit'][0]
    last_day = df['Datum/Zeit'][-1:]
    length = len(df['Datum/Zeit'])
    nans = df.isna().sum().sum()
    df_new = pd.DataFrame({
        'First Day' : first_day,
        'Last Day' : last_day,
        'Length' : length,
        'Count_nan' : nans
    })
    df_results = pd.concat([df_results, df_new])
    #SAVE LIST FOR ANALYSIS
    pollutant_unit.append(pd.read_csv(PATH_STATIONS, encoding='latin-1').iloc[1,0].replace("Einheit: ",""))
    pollutant_kind.append(poll.replace(".csv",""))

df_results
```
